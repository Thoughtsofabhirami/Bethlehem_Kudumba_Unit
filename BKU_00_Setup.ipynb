But a check that changes nothing is only half useful. What should happen <b>next depends on the answer.</b> In Python, a <b>condition</b> lets the next action depend on a check.</p></div><div style="background:#FDEEE6;border-left:8px solid #B65A33;border-radius:12px;padding:14px 18px;margin:6px 0;color:#3B2A20;font-family:'Segoe UI',Roboto,'Helvetica Neue',Arial,sans-serif;line-height:1.6;"><div style="font-weight:700;font-size:1.05em;color:#8E3F22;margin-bottom:4px;">🔍&nbsp; Read first</div><p style="margin:5px 0;">Read the next cell as a sentence: *"If BKU exists, say it's here. Otherwise, say we need to build it."*</p><ul style="margin:4px 0 4px 0;padding-left:22px;"><li style='margin:3px 0;'><code style="background:#FFFFFF;border:1px solid #E5CFB0;border-radius:5px;padding:1px 6px;color:#8E3F22;font-family:Consolas,Menlo,monospace;font-size:.92em;">if</code> means "only when the answer below is <code style="background:#FFFFFF;border:1px solid #E5CFB0;border-radius:5px;padding:1px 6px;color:#8E3F22;font-family:Consolas,Menlo,monospace;font-size:.92em;">True</code>, do the indented lines".</li><li style='margin:3px 0;'><code style="background:#FFFFFF;border:1px solid #E5CFB0;border-radius:5px;padding:1px 6px;color:#8E3F22;font-family:Consolas,Menlo,monospace;font-size:.92em;">os.path.exists(bku_path)</code> is the question.</li><li style='margin:3px 0;'><code style="background:#FFFFFF;border:1px solid #E5CFB0;border-radius:5px;padding:1px 6px;color:#8E3F22;font-family:Consolas,Menlo,monospace;font-size:.92em;">:</code> (the colon) ends the question line, like "then:".</li><li style='margin:3px 0;'>The <b>indented</b> lines (4 spaces in) belong to that branch. ```python
if os.path.exists(bku_path):
    print("The BKU folder is already here. Nothing to build.")
else:
    print("No BKU folder yet. Time to build the house!")
```

<div style="font-family:'Segoe UI',Roboto,'Helvetica Neue',Arial,sans-serif;font-weight:700;color:#8E3F22;font-size:1.15em;margin:10px 0 2px 0;">Step 2.7 &nbsp;Build the house</div><div style="background:#E8F3FB;border-left:8px solid #4F8FC4;border-radius:12px;padding:14px 18px;margin:6px 0;color:#3B2A20;font-family:'Segoe UI',Roboto,'Helvetica Neue',Arial,sans-serif;line-height:1.6;"><div style="font-weight:700;font-size:1.05em;color:#1F4E79;margin-bottom:4px;">📖&nbsp; Story</div><p style="margin:5px 0;">The check said BKU is missing, so now we have a real reason to build. The same <code style="background:#FFFFFF;border:1px solid #E5CFB0;border-radius:5px;padding:1px 6px;color:#8E3F22;font-family:Consolas,Menlo,monospace;font-size:.92em;">if / else</code> shape, but the <code style="background:#FFFFFF;border:1px solid #E5CFB0;border-radius:5px;padding:1px 6px;color:#8E3F22;font-family:Consolas,Menlo,monospace;font-size:.92em;">else</code> branch will now <b>do</b> something.</p></div><div style="background:#FDEEE6;border-left:8px solid #B65A33;border-radius:12px;padding:14px 18px;margin:6px 0;color:#3B2A20;font-family:'Segoe UI',Roboto,'Helvetica Neue',Arial,sans-serif;line-height:1.6;"><div style="font-weight:700;font-size:1.05em;color:#8E3F22;margin-bottom:4px;">🔍&nbsp; Read first</div><p style="margin:5px 0;">The new line is <code style="background:#FFFFFF;border:1px solid #E5CFB0;border-radius:5px;padding:1px 6px;color:#8E3F22;font-family:Consolas,Menlo,monospace;font-size:.92em;">os.makedirs(bku_path, exist_ok=True)</code>:</p><ul style="margin:4px 0 4px 0;padding-left:22px;"><li style='margin:3px 0;'><code style="background:#FFFFFF;border:1px solid #E5CFB0;border-radius:5px;padding:1px 6px;color:#8E3F22;font-family:Consolas,Menlo,monospace;font-size:.92em;">makedirs</code> means <b>make</b> <b>dir</b>ector<b>ies</b> (folders). If some folders on the way are missing, it makes those too.</li><li style='margin:3px 0;'>It has <b>two arguments</b>, separated by a comma. The first is the address to build.</li><li style='margin:3px 0;'><code style="background:#FFFFFF;border:1px solid #E5CFB0;border-radius:5px;padding:1px 6px;color:#8E3F22;font-family:Consolas,Menlo,monospace;font-size:.92em;">exist_ok=True</code> is a <b>named argument</b>: we write the setting's name, then <code style="background:#FFFFFF;border:1px solid #E5CFB0;border-radius:5px;padding:1px 6px;color:#8E3F22;font-family:Consolas,Menlo,monospace;font-size:.92em;">=</code>, then its value. Without <code style="background:#FFFFFF;border:1px solid #E5CFB0;border-radius:5px;padding:1px 6px;color:#8E3F22;font-family:Consolas,Menlo,monospace;font-size:.92em;">exist_ok=True</code>, running it twice would crash on the second run. ```python
if os.path.exists(bku_path):
    print("The BKU folder is already here. Nothing to build.")
else:
    os.makedirs(bku_path, exist_ok=True)
    print("The BKU house is built at:", bku_path)
```

```python
#@title 🔒 The reward lock  (run once; nothing to read or fill in here yet) { display-mode: "form" }
import base64, hashlib, urllib.request
from IPython.display import display, HTML

try:
    from cryptography.fernet import Fernet, InvalidToken
except ImportError:
    import subprocess, sys
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "cryptography"])
    from cryptography.fernet import Fernet, InvalidToken

# ---- set once by the notebook's creator ------------------------------
REWARD_REPO   = "YOUR-GITHUB-NAME/YOUR-REPO"
REWARD_BRANCH = "main"
REWARD_FOLDER = "rewards"
# ----------------------------------------------------------------------

def _make_key(answer):
    # The key is made from the learner's answer: tidy it, then scramble it.
    text = str(answer).strip().lower().rstrip("/")
    raw = hashlib.pbkdf2_hmac("sha256", text.encode("utf-8"), b"BKU-reward-lock-v1", 200000)
    return base64.urlsafe_b64encode(raw)


def _card(emoji, title, line):
    display(HTML(
        '<div style="border:3px dashed #B65A33;background:#FBF3E4;color:#8E3F22;border-radius:14px;'
        'padding:26px 18px;text-align:center;font-family:Segoe UI,Arial,sans-serif;">'
        f'<div style="font-size:2.2em;">{emoji}</div>'
        f'<div style="font-size:1.15em;font-weight:700;">{title}</div>'
        f'<div style="margin-top:6px;">{line}</div></div>'))


def _mime(data):
    if data[:3] == b"\xff\xd8\xff": return "image/jpeg"
    if data[:8] == b"\x89PNG\r\n\x1a\n": return "image/png"
    if data[:4] == b"RIFF" and data[8:12] == b"WEBP": return "image/webp"
    if data[:3] == b"GIF": return "image/gif"
    return "image/jpeg"


def unlock_reward(scene, answer, hint="Check each blank again, then press Shift + Enter.", caption=""):
    if REWARD_REPO.startswith("YOUR-GITHUB-NAME"):
        _card("🎬", "REWARD IMAGE GOES HERE", "The reward chest is not connected yet.")
        return
    url = f"https://raw.githubusercontent.com/{REWARD_REPO}/{REWARD_BRANCH}/{REWARD_FOLDER}/{scene}.bin"
    try:
        with urllib.request.urlopen(url, timeout=20) as reply:
            locked = reply.read()
    except Exception:
        _card("📡", "Could not reach the reward chest", "Check your internet connection, then run this cell again.")
        return
    try:
        picture = Fernet(_make_key(answer)).decrypt(locked)
    except InvalidToken:
        _card("🔒", "The lock did not open yet", hint)
        return
    data = base64.b64encode(picture).decode()
    words = f'<div style="margin-top:8px;font-weight:600;">{caption}</div>' if caption else ""
    display(HTML(
        '<div style="background:#FBF3E4;border:3px solid #B65A33;border-radius:16px;padding:14px;'
        'text-align:center;font-family:Segoe UI,Arial,sans-serif;color:#8E3F22;'
        'box-shadow:0 4px 14px rgba(0,0,0,.2);">'
        '<div style="font-weight:800;letter-spacing:2px;margin-bottom:8px;">🌺 THE CHEST OPENS 🌺</div>'
        f'<img src="data:{_mime(picture)};base64,{data}" style="max-width:100%;border-radius:10px;">'
        f'{words}</div>'))
```
